# Task
Translate the "input.wav" audio file into English using the Whisper model and display the translated text.

## Load Whisper Model

### Subtask:
Load the pre-trained Whisper model.


**Reasoning**:
To load the pre-trained Whisper model, I need to import the `whisper` library and then use `whisper.load_model()` to load a model, storing it in a variable named `model`.



```python
import whisper

model = whisper.load_model("base")
print("Whisper model 'base' loaded successfully.")
```

**Reasoning**:
Now that the Whisper model is loaded, I will use it to transcribe the `input.wav` audio file. The `transcribe` method will perform the translation, and I will print the resulting text.



```python
result = model.transcribe("input.wav", task="translate")
print(result["text"])
```

# Task
Save the translated text from `result['text']` to a file named 'translation.txt'.

## Save Translation to File

### Subtask:
Save the translated text from the Whisper model into a file named 'translation.txt'.


**Reasoning**:
I need to save the translated text, which is stored in `result['text']`, to a file named 'translation.txt'. I will open the file in write mode, write the text, and then close the file.



```python
with open('translation.txt', 'w') as f:
    f.write(result['text'])
print("Translation saved to 'translation.txt'.")
```
